### Prompt

Mögliche Prompts siehe [hier](https://gitlab.com/ch-mc-b/autoshop-ms/edu/chatgpt)

```python
import os
os.environ['OPENAI_QUESTION']='''
Basierend auf den Microservices Beispiel von Eberhard Wolff schreibe mit 3 Python Scripte welche jeweils eine Webseite mit
- Kundendaten (Customer)
- Produktdaten (Catalog)
- Bestellungen (Order)
und ein Einsteigerscript in Python mit einem Menü für den Wechsel auf Customer, Catalog und Order enthält.
Die Script sollen ein REST-API /<service>/api und die Daten im HTML-Format als /<service/ zur Verfügung stellen.
Die Produkte sollen dem eines Auto Hauses mit Autos und Auto Zubehör Artikeln entsprechen.

Als TCP/IP Port soll generell 8080 und die Services sollen gegen 0.0.0.0 geöffnet sein.
Das Einsteigerscript soll das Reverse Proxy Design Pattern verwenden.

'''
```

```python
from openai import OpenAI
import os

client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))
question = os.getenv("OPENAI_QUESTION")

# Anfrage an OpenAI senden
response = client.chat.completions.create(
    model="gpt-4.1-nano",  # alternativ: "gpt-4-turbo" oder "gpt-3.5-turbo"
    messages=[
        {"role": "user", "content": question}
    ]
)

# Antwort ausgeben
print(response.choices[0].message.content)
```


**Anmerkung**: Es wäre Besser das Response API zu verwenden, dieses kennt die bereits gestellten Fragen.

```python
from openai import OpenAI
import os

# Initialize OpenAI client with your API key
client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))

# Load the question from environment variable
question = os.getenv("OPENAI_QUESTION")

# Make the API call
response = client.chat.completions.create(
    model="gpt-4.1-nano",  # you can also use "gpt-4-turbo" or "gpt-3.5-turbo"
    messages=[
        {"role": "user", "content": question}
    ]
)

# The `response` object is a structured Response
# We can safely access the content like this:

answer = response.choices[0].message.content

# Print the answer
print(answer)

# Optionally: Save to a Markdown file
with open("answer.md", "w") as f:
    f.write(answer)
```
